## 13. nunique()

## 14. value_counts()

The `value_counts()` method counts how many times each unique value occurs in a column.

It is mainly useful for analyzing categorical variables and understanding the distribution of their categories.

```python
# Count the values of the target variable

df["Churn"].value_counts()
```

### Observation

The `Churn` column contains two categories:

- `No`
- `Yes`

The value counts show how many customers belong to each category.

This is important because it helps us understand the distribution of the target variable and identify whether the classes are balanced or imbalanced.

```python
# Calculate the percentage distribution of the target variable

df["Churn"].value_counts(normalize=True) * 100
```

### Inspecting Other Categorical Variables

`value_counts()` can also be used to understand the distribution of individual categorical features.

```python
# Distribution of Contract categories

df["Contract"].value_counts()
```

```python
# Distribution of InternetService categories

df["InternetService"].value_counts()
```

## 15. Column Classification

After inspecting the dataset, the columns can be classified into different groups:

- Numerical columns
- Categorical columns
- Date/time columns
- Identifier columns
- Target column

This classification helps determine the appropriate analysis and preprocessing techniques for each type of variable.

```python
# Numerical columns

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

# Categorical/object columns

categorical_columns = df.select_dtypes(include="object").columns.tolist()

# Identifier column

identifier_columns = ["customerID"]

# Target column

target_column = ["Churn"]

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical/Object Columns:")
print(categorical_columns)

print("\nIdentifier Column:")
print(identifier_columns)

print("\nTarget Column:")
print(target_column)
```
