# 1. Normal Distribution

## Concept

A Normal Distribution is a continuous probability distribution that has a bell-shaped curve.

Most values are concentrated around the center, and fewer values occur as we move away from the center.

A normal distribution is symmetric around its mean.

## Shape

The distribution looks like a bell.

The left and right sides are approximately equal.

## Center

For a perfectly normal distribution:

Mean = Median = Mode

## Spread

The spread is mainly controlled by the standard deviation.

A larger standard deviation means the data is more spread out.

## Real-World Example

Human heights in a large population can approximately follow a normal distribution.

## AI/ML Example

Many statistical methods and ML techniques use assumptions related to normally distributed features.

## Important Point

A normal distribution does not automatically mean that there are no outliers. Outliers can still exist.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

The histogram should show a bell-shaped distribution.

The center of the data is around 50.

Most observations are close to the center, while fewer observations appear at the extremes.

# 2. Uniform Distribution

## Concept

In a Uniform Distribution, values within a particular range have approximately equal probability of occurring.

Unlike a normal distribution, there is no central peak.

## Shape

The distribution looks approximately flat.

## Center

For a continuous uniform distribution between `a` and `b`, the center is:

(a + b) / 2

## Spread

The spread depends on the distance between the minimum and maximum values.

## Real-World Example

A random number selected between 1 and 10 can follow a uniform distribution.

## AI/ML Example

Uniform random values can be useful when generating random data for simulations and testing algorithms.

```python
uniform_data = np.random.uniform(
    low=1,
    high=10,
    size=1000
)

plt.hist(uniform_data, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

The histogram should appear relatively flat.

Values between 1 and 10 have approximately similar chances of appearing.

# 3. Skewness

## Concept

Skewness tells us how much a distribution is asymmetric.

It helps us understand whether the data has a longer tail on the left or right side.

There are three common cases:

- Skewness ≈ 0 → approximately symmetric
- Positive skewness → right-skewed
- Negative skewness → left-skewed

## Why It Matters

Skewness helps us understand the shape of numerical data and identify whether extreme values may be affecting the distribution.

## AI/ML Example

Highly skewed features such as income or transaction amounts may sometimes require transformation before being used in a machine learning model.

```python
from scipy.stats import skew

data = [10, 12, 13, 14, 15, 16, 20]

skewness_value = skew(data)

print("Skewness:", skewness_value)
```

## Interpretation

The skewness value tells us about the asymmetry of the distribution.

A value close to zero indicates that the distribution is approximately symmetric.

A positive value indicates right skewness, while a negative value indicates left skewness.

# 4. Kurtosis

## Concept

Kurtosis describes the shape of the tails of a distribution and how strongly extreme values occur compared with a normal distribution.

It helps us understand the tendency of data to contain extreme observations.

## Interpretation

Higher kurtosis generally indicates heavier tails and a greater tendency for extreme values.

Lower kurtosis generally indicates lighter tails.

## Real-World Example

Financial returns can sometimes have high kurtosis because extreme changes may occur more often than expected under a normal distribution.

## AI/ML Example

Kurtosis can help during exploratory data analysis to identify features that may contain extreme observations.

```python
from scipy.stats import kurtosis

data = [10, 12, 13, 14, 15, 16, 20, 50]

kurtosis_value = kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

## Interpretation

The kurtosis value gives information about the tails of the distribution.

A higher value can indicate that extreme values are more influential in the dataset.

# 5. Symmetric Distribution

## Concept

A distribution is symmetric when the left side and right side have approximately the same shape around the center.

If we draw a vertical line through the center, both sides look similar.

## Center

In a perfectly symmetric distribution, the mean and median are generally located at the center.

## Example

A normal distribution is a common example of a symmetric distribution.

## AI/ML Example

Checking symmetry helps us understand whether numerical features have strong skewness before applying statistical or machine learning techniques.

```python
symmetric_data = np.random.normal(50, 8, 1000)

plt.hist(symmetric_data, bins=30)
plt.title("Symmetric Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

The graph should look approximately balanced on both sides of the center.

There should not be a long tail on only one side.

# 6. Left-Skewed Distribution

## Concept

A left-skewed distribution has a longer tail toward the left side.

It is also called a negatively skewed distribution.

In many left-skewed datasets:

Mean < Median

## Real-World Example

Exam scores can sometimes be left-skewed when most students perform very well but a few students score very low.

## AI/ML Example

Understanding left skewness helps us identify unusual values and understand the distribution of a numerical feature.

```python
left_skewed = 100 - np.random.exponential(
    scale=10,
    size=1000
)

plt.hist(left_skewed, bins=30)
plt.title("Left-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

The distribution should have most of its values toward the right side and a longer tail extending toward the left.

This indicates negative or left skewness.

# 7. Right-Skewed Distribution

## Concept

A right-skewed distribution has a longer tail toward the right side.

It is also called a positively skewed distribution.

In many right-skewed datasets:

Mean > Median

## Real-World Example

Income is often right-skewed because most people may have moderate incomes while a small number of people have extremely high incomes.

## AI/ML Example

Transaction amounts, income, and property prices can contain right-skewed features. Understanding the skewness can help during preprocessing and feature transformation.

```python
right_skewed = np.random.exponential(
    scale=10,
    size=1000
)

plt.hist(right_skewed, bins=30)
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

Most values should be concentrated toward the left, while a longer tail extends toward the right.

This indicates positive or right skewness.

# 8. Distribution Analysis of Numerical Features

## Objective

Now we will analyze numerical features from a dataset.

For each numerical feature, we will examine:

- Shape of distribution
- Center
- Spread
- Skewness
- Presence of outliers

This helps us understand the data before applying machine learning algorithms.

```python
df = pd.DataFrame({
    "Age": [20, 21, 22, 25, 24, 23, 26, 27, 22, 21],
    "Salary": [25000, 28000, 30000, 32000, 35000,
               40000, 45000, 50000, 55000, 100000],
    "Marks": [70, 75, 80, 82, 85, 88, 90, 92, 95, 98]
})

df
```

```python
numerical_features = df.select_dtypes(
    include=np.number
).columns

print(numerical_features)
```

## Center and Spread

We can use the mean and standard deviation to understand the center and spread of numerical features.

```python
print("Mean:")
print(df[numerical_features].mean())

print("\nStandard Deviation:")
print(df[numerical_features].std())
```

## Skewness of Features

We can calculate skewness for every numerical feature to understand whether the feature is symmetric, left-skewed, or right-skewed.

```python
print(df[numerical_features].skew())
```

## Checking for Outliers

One simple method to identify possible outliers is the IQR method.

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries can be considered potential outliers.

```python
for column in numerical_features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

# Box Plot

A box plot helps us understand the center, spread, and potential outliers of numerical data.

It shows:

- Median
- Quartiles
- IQR
- Potential outliers

This makes it useful for quickly identifying unusual values.

```python
plt.boxplot(df["Salary"])
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

## Interpretation

The box plot helps us see the middle 50% of the data through the box.

Points outside the whiskers may represent potential outliers.

In the Salary feature, the value 100000 may appear as an unusually high value compared with the other salaries.
