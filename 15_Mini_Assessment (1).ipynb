## Question 1

What is EDA?

### Answer

EDA stands for Exploratory Data Analysis.

It is the process of examining and understanding a dataset before applying statistical analysis or machine learning.

EDA helps us understand the structure of the data, identify missing values and duplicates, detect outliers, study distributions, find relationships between variables, and discover important patterns.

## Question 2

Why is EDA performed before Machine Learning?

### Answer

EDA is performed before Machine Learning to understand the quality and characteristics of the dataset.

It helps identify problems such as missing values, duplicate records, incorrect data types, outliers, class imbalance, and unusual distributions.

EDA also helps identify relationships between features and the target variable.

This allows us to prepare better data and make more informed decisions during the Machine Learning process.

## Question 3

What is the difference between univariate, bivariate, and multivariate analysis?

### Answer

Univariate analysis studies one variable at a time.

Example:
Analyzing the distribution of Salary.

Bivariate analysis studies the relationship between two variables.

Example:
Analyzing the relationship between Salary and Experience.

Multivariate analysis studies three or more variables at the same time.

Example:
Analyzing Salary, Experience, Age, and Department together.

## Question 4

How do you identify missing values in a dataset?

### Answer

Missing values can be identified using Pandas.

The `isnull()` function checks whether values are missing, and `sum()` counts the missing values in each column.

For example:

`df.isnull().sum()`

This shows the number of missing values in every column.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, None, 30, 35],
    "Salary": [30000, 40000, 50000, None, 60000]
})

print(df.isnull().sum())
```

## Question 5

How do you identify duplicate records in a dataset?

### Answer

Duplicate records can be identified using the Pandas `duplicated()` function.

The `sum()` function can then be used to count the number of duplicate rows.

For example:

`df.duplicated().sum()`

returns the total number of duplicate rows.

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C", "B"],
    "Age": [22, 25, 30, 25]
})

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

## Question 6

What is an outlier?

### Answer

An outlier is a data value that is unusually far away from most of the other observations.

For example, if most employee salaries are between 30,000 and 80,000 but one salary is 500,000, that value may be an outlier.

Outliers can occur because of genuine unusual observations, measurement errors, or data-entry errors.

## Question 7

Explain the IQR method for detecting outliers.

### Answer

IQR stands for Interquartile Range.

It measures the distance between the first quartile (Q1) and third quartile (Q3).

Formula:

IQR = Q3 - Q1

The lower boundary is:

Q1 - 1.5 × IQR

The upper boundary is:

Q3 + 1.5 × IQR

Values below the lower boundary or above the upper boundary are considered potential outliers.

```python
import numpy as np

data = np.array([10, 12, 14, 15, 16, 18, 20, 22, 25, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_bound) | (data > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

- Positively skewed or right-skewed
- Negatively skewed or left-skewed
- Approximately symmetric

In a left-skewed distribution, the tail extends toward smaller values.

## Question 9

What is correlation?

### Answer

Correlation measures the strength and direction of the relationship between two variables.

A positive correlation means that the variables tend to increase together.

A negative correlation means that when one variable increases, the other tends to decrease.

A correlation close to zero indicates little or no linear relationship.

## Question 10

What is the difference between correlation and causation?

### Answer

Correlation means that two variables are related or change together.

Causation means that a change in one variable directly causes a change in another variable.

Correlation does not necessarily mean causation.

For example, ice cream sales and swimming activity may both increase during summer. They are correlated, but buying ice cream does not cause people to go swimming.

## Question 11

What is multicollinearity?

### Answer

Multicollinearity occurs when two or more independent features in a dataset are highly related to each other.

For example, if a dataset contains `Age` and `Years_of_Experience`, these variables may be strongly related.

Multicollinearity can make it difficult to determine the individual effect of each feature, especially in models such as Linear Regression.

## Question 12

How do you identify class imbalance?

### Answer

Class imbalance occurs when the classes in a classification target variable are not represented equally.

It can be identified by counting the number of observations in each class.

For example, if a dataset contains 9,000 customers who did not churn and only 1,000 customers who churned, the target variable is imbalanced.

Pandas `value_counts()` can be used to examine the class distribution.

```python
import pandas as pd

churn = pd.Series([
    "No", "No", "No", "No", "No",
    "No", "No", "Yes", "No", "Yes"
])

print(churn.value_counts())

print("\nClass Percentage:")
print(churn.value_counts(normalize=True) * 100)
```

## Question 13

Why should outliers not always be removed?

### Answer

Outliers should not always be removed because they may represent genuine and important observations.

For example, a very high salary may belong to a senior executive and therefore be a valid value.

Removing valid outliers can cause loss of useful information and may make the dataset less representative.

Outliers should first be investigated before deciding whether they should be removed, transformed, or retained.

## Question 15

Detect outliers using the IQR method.

```python
import numpy as np

data = np.array([
    10, 12, 13, 14, 15,
    16, 17, 18, 20, 100
])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_bound) |
    (data > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

## Question 16

Analyze the distribution of a numerical feature.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Salary": [
        30000, 32000, 35000, 37000, 40000,
        42000, 45000, 48000, 50000, 55000,
        60000, 65000, 70000
    ]
})

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Salary",
    kde=True
)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()

print("Mean Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
```

## Question 17

Perform categorical analysis on a categorical feature and display the frequency of each category.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Department": [
        "IT", "HR", "IT", "Finance",
        "IT", "HR", "Finance", "IT",
        "HR", "Finance"
    ]
})

print("Category Frequency:")
print(df["Department"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Department"
)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

## Question 18

Generate a complete EDA summary for a dataset.

### Includes:

- Dataset shape
- Data types
- Missing values
- Duplicate records
- Numerical summary
- Categorical analysis
- Outlier analysis
- Correlation analysis
- Important observations

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35, 40, 45, 50, 55, 60],
    "Salary": [30000, 35000, 40000, 45000, 50000, 60000, 70000, 80000, 90000, 100000],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "IT", "HR", "Finance"
    ]
})

# Dataset shape
print("Dataset Shape:")
print(df.shape)

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Numerical summary
print("\nNumerical Summary:")
print(df.describe())

# Categorical analysis
print("\nDepartment Frequency:")
print(df["Department"].value_counts())

# Outlier analysis
print("\nOutlier Analysis:")

for column in ["Age", "Salary"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(column, "outliers:", len(outliers))

# Correlation analysis
print("\nCorrelation Matrix:")
print(df[["Age", "Salary"]].corr())
```
